```python
import pandas as pd
from sodapy import Socrata
from datetime import datetime, timedelta
```

```python
client = Socrata("www.dallasopendata.com", None)
```

```python
yesterday = datetime.now() - timedelta(days=1)
date = yesterday.date()
string = "ararrestdate = '" + str(date) + "'"
results = client.get("sdr7-6v3j", where=string)
```

```python
results_df = pd.DataFrame.from_records(results)
results_df.head(5)
```

```python
results_df.to_csv("arrest_apr_1st_code.csv", index=False)
```
